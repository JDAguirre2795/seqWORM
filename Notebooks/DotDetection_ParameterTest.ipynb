## Dot Detection Parameter Testing
Test preprocessing + dot detection parameters on a few stacks before the full run.
Typical use: one positive sample and one no-probe control. The no-probe control should give (close to) zero spots.

For each channel the figure shows the preprocessed MIP next to the same MIP with detected dots, full field and a zoom on the brightest region.
Contrast is shared across samples per channel, so intensities are directly comparable.
Preprocessing is done in memory, so preprocessing parameters can be tested too.
