## Preprocessing
Background correction and low-pass filtering of smFISH stacks before dot detection.
Port of the pyfish_tools `pre_processing.py` workflow (`JD_Preprocessing.ipynb`):
0. Camera hot pixels (single pixels bright in every plane) are found per stack, saved as `<name>_hotpixels.csv`, and replaced by the median of their 8 neighbours
1. FISH channels: subtract Gaussian-blurred background (kern=9, sigma=6), then gamma correction (1.6)
2. Last channel (DAPI): radial edge-dimming correction (least-squares)
3. All channels: low-pass Gaussian blur (kern=9, sigma=1)

Expects (Z, C, Y, X) TIFF stacks, e.g. from `IMS to TIFF conversion.ipynb`. 2D images are skipped.
