## Process all folders of TIFFs
This will scan all subfolders in the root folder and convert every TIFF file into an H5 file.


```python
# Set root folder
root_folder = r"C:\Users\josed\GitHub\WormFISH\data"  # Change this to your actual folder path

# Process folders
created_files = process_folders(root_folder)
print(f"Total H5 files created: {len(created_files)}")
for f in created_files:
    print(f)
```

## Loop through all positions - Merge H5 files into a single dataset per position

Use this if all your dimensions are the same across all your files.
This will create a merged H5 dataset with all points and frames for each position within a root directory.


```python
merge_all_positions(
    root_folder=root_folder,  # Path to the folder containing 'raw H5 files'
    N_points=200,             # Number of points to be added for annotation
    C=3,                      # Number of channels
    W=2048,                   # Image width
    H=2048,                   # Image height
    D=77,                     # Depth/Z dimension
    clip_quantiles=(0, 1)     # Percentiles for clipping 16-bit images
)
```

## Single position processing- Merge H5 files into a single dataset fo

Use this if your dimensions are different across positions
This will create a merged H5 dataset with all points and frames for a unique position


```python
pos_number = "1"  # or "2", "3", etc.
pos_folder = os.path.join(root_folder, "raw H5 files", f"pos{pos_number}")
merged_output = os.path.join(root_folder, "MergedH5", f"mergedpos{pos_number}.h5")

merge_h5_files(
    directory=pos_folder,
    output_file=merged_output,
    N_points=200, #Number of points to be added for annotation
    C=3, #Number of channels
    W=2048, #Image widht
    H=2048, #Image height
    D=77, #Depth/Z dimension
    clip_quantiles=(0, 1) #Percentiles for cliping 16-bit images
)
```
