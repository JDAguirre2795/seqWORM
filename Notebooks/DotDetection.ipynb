`z_trim` removes blank padding planes and the first/last planes of each stack (edge debris and out-of-focus light).
Saves `spots_<channel>_<file>.npy` (after decomposition), `spots_<channel>_detected_<file>.npy` (before decomposition) and `spot_counts.csv`.
Thresholds 100 (A647) / 200 (CY3B) and decomposition alpha 0.7 / beta 2.5 were chosen with no-probe controls (`DotDetection_ParameterTest.ipynb`).
Results are saved in a DetectedDots folder within the root directory.
